Sometimes I want to put the value of a variable into a markdown cell. I know of three ways to inject variable values in Jupyter Lab markdown:

1. Using the [IPython `Markdown` display object](https://ipython.readthedocs.io/en/8.26.0/api/generated/IPython.display.html#IPython.display.Markdown) and [f-strings](https://docs.python.org/3/tutorial/inputoutput.html#formatted-string-literals). [⬇️](#sec-ipythondisplay)
2. If rendering a notebook with Quarto, you can use the Quarto [inline code](https://quarto.org/docs/computations/inline-code.html) syntax. [⬇️](#sec-quartoinline)
3. Using the `jupyterlab-myst` plugin/syntax. (Not included below, see [docs](https://mystmd.org/guide/quickstart-jupyter-lab-myst))


## IPython display Markdown object {#sec-ipythondisplay}

````python
dynamic_content = random.randint(0, 100)

Markdown(f"""This is the dynamic content in bold: **{dynamic_content=}**
""")
````

You could hide the code from a document rendered by Quarto with by adding an `#| echo: false` comment to the code cell.

## Quarto inline code syntax {#sec-quartoinline}

::: {.callout-important}
You must turn on notebook execution when rendering with quarto for the following to work. 

E.g. you can add, 
``` markdown
execute: 
    enabled: true 
```
to your notebook front matter.
:::

The Quarto syntax for injecting the value of a variable into markdown content looks like this:


```markdown
This is the dynamic content in bold **`{{python}} f'{dynamic_content=}'`**
```

And here ⬇️ is what the output looks like when I actually use that syntax in a markdown cell.

> This is the dynamic content in bold **`{python} f'{dynamic_content=}'`**
